```python
from qiskit import QuantumCircuit
```

```python
qc = QuantumCircuit(1)
qc.x(0)
qc.draw('mpl')
```

```python
qc1 = QuantumCircuit(3)
qc1.ccx(0,1,2)
qc1.draw('mpl')
```

```python
qc2 = QuantumCircuit(3)
qc2.x([0,1,2])
qc2.ccx(0,1,2)
qc2.x([0,1])
qc2.draw('mpl')
```

```python
qc3 = QuantumCircuit(3)
qc3.cx(0,2)
qc3.cx(1,2)
qc3.draw('mpl')
```

```python
qc4 = QuantumCircuit(3)
qc4.initialize('1',2)
qc4.ccx(0,1,2)
qc4.draw('mpl')
```

```python
qc5 = QuantumCircuit(3)
qc5.x([0,1])
qc5.ccx(0,1,2)
qc5.x([0,1])
qc5.draw('mpl')
```
